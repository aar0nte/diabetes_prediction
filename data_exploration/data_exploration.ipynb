# Jupyter Notebook for data exploration

Importing libraries

```python
import pandas as pd
import numpy as np
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

Import data and sot out colums where Glucose, Blood Pressure, Skin Thickness and BMI are 0

```python
data = pd.read_csv('../data/diabetes.csv')

print(f"Columns with 0 before ereasure: \n{data[['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']].eq(0).sum()}\n")

data.drop('Insulin', axis=1, inplace=True) # Wieso habe ich hier Insulin gedroppt?

cols_with_zero = ['Glucose', 'BloodPressure', 'SkinThickness', 'BMI']

for col in cols_with_zero:
    median_value = data[col].median()
    data[col] = data[col].replace(0, median_value)

print(f"Columns with 0 after ereasure: \n{data[['Glucose', 'BloodPressure', 'SkinThickness', 'BMI']].eq(0).sum()}")

data
```

Some tiny calculations

```python
count_positive = (data["Outcome"] == 1).sum()
print(f"The number of positive diabetes cases of this data set is n = {count_positive}.")

ratio_positive = count_positive / data.shape[0] * 100
print(f"This is {count_positive / data.shape[0] * 100:.2f}% of the all patients of the dataset.")
```

```python
X = data.drop(["Outcome"], axis = 1)
y = data["Outcome"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = .2, random_state = 161)
```

## Using NeighborsClassifier:

```python
for n in range(1,50):
    clf = KNeighborsClassifier(n_neighbors = n)
    clf.fit(X_train, y_train)
    predictions = clf.predict(X_test)

    print(f"accuracy score with {n} neighbors: {accuracy_score(predictions, y_test)}")
```
